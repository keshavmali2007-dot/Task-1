```python
import pandas as pd
import numpy as np
import matplotlib.pylab as plt
import seaborn as sns
plt.style.use('ggplot')
```

```python
df = pd.read_csv( '../input/training-data/atlantis_citizens_final.csv') 
```

**QUESTION 1**

```python
df["Person_working_out"] = df["District_Name"] != df["Work_District"]
```

```python
Person_CommuteOut = df.groupby("District_Name")["Person_working_out"].mean()
```

```python
Percentage = Person_CommuteOut.mul(100)
```

```python
Percentage = Percentage.reset_index()
```

```python
Percentage
```

```python
Percentage.to_csv("/kaggle/working/Percentage_CommuteOut.csv")
```

```python
Occupation = df.groupby(["District_Name" , "Occupation"])["Person_working_out"].mean()
```

```python
Occupation = Occupation.mul(100)
```

```python
Occupation_Percentage = Occupation.reset_index()
```

```python
Occupation_Percentage
```

```python
Occupation_Percentage.to_csv("/kaggle/working/Occupation_CommuteOut.csv")
```
